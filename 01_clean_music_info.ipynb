```python
import pandas as pd

print(pd.__version__)
```

```python
df = pd.read_csv("Music Info.csv")
print(df.shape)
print(df.isnull().sum())
```

```python
df_clean = df[df["tempo"] != 0]
print(df.shape)
print(df_clean.shape)
```

```python
df_clean["genre"] = df_clean["genre"].fillna("Unkown")
print(df_clean["genre"].isnull().sum())
audio_features = [
    "danceability",
    "energy",
    "loudness",
    "speechiness",
    "acousticness",
    "instrumentalness",
    "liveness",
    "valence",
    "tempo",
]
X = df_clean[audio_features]
print(X.shape)
print(X.head())
X.describe()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled_df = pd.DataFrame(X_scaled, columns=audio_features)
X_scaled_df.describe()
```

```python
# the cleaned dataframe (with genre='Unknown' filled, tempo=0 rows dropped)
df_clean.to_csv("music_info_clean.csv", index=False)

# the scaled feature matrix, ready for k-means
import numpy as np

np.save("X_scaled.npy", X_scaled)

import joblib

joblib.dump(scaler, "scaler.pkl")
```
